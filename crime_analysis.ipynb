## 1) Crime Trend By Year

```python
year = df.groupby("Crime_Year").size()
year.plot(kind="bar")
plt.xlabel("Year")
plt.ylabel("Number of Crimes")
plt.title("Crime Trend By Year")
plt.show()
```

Crime remained relatively stable from 2020 to 2024,with noticebale fluctuations in certain years. This indicates periods where targeted policing strategies may have influenced crime patterns.

## 2) Top 10 Crime Types

```python
top = df["Crm_Cd_Desc"].value_counts().head(10).sort_values(ascending=True)
top.plot(kind = "barh")
plt.title("Top 10 Crime Types")
plt.ylabel("Crime Type")
plt.xlabel("Count")
plt.show()
```

## 3) Top Crime Areas

```python
areas = df["AREA_NAME"].value_counts().sort_values(ascending=True).head(10)
areas.plot(kind = "barh")
plt.title("Crime By Area")
plt.ylabel("Area")
plt.xlabel("Count")
plt.show()
```

A small number of police areas contribute a disproportionately high share of reported crimes,indicating potential hotspots for resource allocation.

## 4) Victim Age Distribution

```python
age = df["Vict_Age"]
age.hist(bins = 20)
plt.title("Victim Age Distribution")
plt.show()
```

## 5) Victim Age Groups

```python
age_group = df["Age_Group"].value_counts()
age_group.plot(kind = "bar")
plt.show()
```

## 6) Victim Gender

```python
gender = df["Vict_Sex"].value_counts()
gender.plot(kind = "pie",autopct="%1.1f%%")
plt.show()
```

## 7) Weapon Usage

```python
weapon = df["Weapon_Desc"].value_counts().head(10)
weapon.plot(kind = "bar")
plt.show()
```

## 8) Monthly Trend

```python
monthly = df.groupby("Crime_Month").size()
monthly.plot(kind ="line",marker = 'o')
plt.show()
```

## 9) Crime Hour Distribution

```python
sns.histplot(df["Crime_Hour"],bins=20)
plt.show()
```

Crime incidents peak during the evening hours,suggesting that police patrols and surveillance could be strengthened during these periods.

## 10) Crime By Quarter

```python
quarter = df.groupby("Quarter_No").size()
quarter.plot(kind='bar')
plt.title("Quarter Wise Crimes")
plt.show()
```

## 17) Top Crime Locations

## 18) Average Victim Age By Crime Type

```python
avg = df.groupby("Crm_Cd_Desc")["Vict_Age"].mean().sort_values(ascending=True).head(10)
avg.plot(kind = "bar")
plt.show()
```

## 19) Crime Heatmap By Month And Year

```python
pivot = pd.pivot_table(
    df,
    index="Crime_Year",
    columns="Month_Name",
    values="DR_NO",
    aggfunc="count"
)

sns.heatmap(pivot)

plt.show()
```
